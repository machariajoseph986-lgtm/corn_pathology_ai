### Image Preprocessing Methodology

Image preprocessing is required to convert the heterogeneous source images into a consistent representation suitable for deep learning.

First, all images are converted to RGB to ensure that the model receives three consistent color channels regardless of the original image color mode.

Second, images are resized to the input dimensions required by each model. V1 uses 256 × 256 pixels, while V2 and V3 use 320 × 320 pixels.

Third, pixel-value normalization is applied according to the model architecture. The custom CNN experiments (V1 and V2) scale pixel values from the original [0, 255] range to [0, 1]. The MobileNetV2 experiment (V3) uses the preprocessing function provided for MobileNetV2, producing values in approximately the [-1, 1] range.

Data augmentation is applied only to the training data. This increases variation in the training examples while keeping the validation and test sets unchanged for objective evaluation.

The validation and test datasets therefore represent unseen data and are not augmented. This separation helps ensure that reported model performance reflects generalization rather than exposure to artificially modified evaluation images.

````python
# PURPOSE: Create a reusable function for RGB conversion, resizing, and array conversion.
import numpy as np

def prepare_image(image_path, target_size=IMG_SIZE_V1):
    """
    Load an image and standardize it for model input.

    Steps:
    1. Load the image.
    2. Convert it to RGB.
    3. Resize it to the requested target size.
    4. Convert it to a NumPy array.
    """

    image = Image.open(image_path).convert("RGB")
    image = image.resize(target_size)

    image_array = np.asarray(image, dtype=np.float32)

    return image_array


# Verify the reusable function using the representative image.
prepared_sample = prepare_image(
    sample_image_path,
    target_size=IMG_SIZE_V1
)

print("REUSABLE IMAGE PREPROCESSING FUNCTION")
print("=" * 80)

print(f"Output shape: {prepared_sample.shape}")
print(f"Output dtype: {prepared_sample.dtype}")
print(f"Pixel minimum: {prepared_sample.min():.1f}")
print(f"Pixel maximum: {prepared_sample.max():.1f}")

assert prepared_sample.shape == (256, 256, 3)
assert prepared_sample.dtype == np.float32
assert 0 <= prepared_sample.min() <= 255
assert 0 <= prepared_sample.max() <= 255

print("\nSTATUS: Reusable image preparation function verified.")
````

````python
# PURPOSE: Verify the V1 pixel normalization process that scales image values from [0, 255] to [0, 1].

# Normalize the prepared sample for V1.
v1_normalized_sample = prepared_sample / 255.0

print("V1 PIXEL NORMALIZATION VERIFICATION")
print("=" * 80)

print(f"Before normalization:")
print(f"  Minimum pixel value: {prepared_sample.min():.4f}")
print(f"  Maximum pixel value: {prepared_sample.max():.4f}")

print("\nAfter normalization:")
print(f"  Minimum pixel value: {v1_normalized_sample.min():.4f}")
print(f"  Maximum pixel value: {v1_normalized_sample.max():.4f}")

assert v1_normalized_sample.dtype == np.float32
assert v1_normalized_sample.min() >= 0.0
assert v1_normalized_sample.max() <= 1.0

print("\nSTATUS: V1 normalization verified.")
````

````python
# PURPOSE: Verify the MobileNetV2 preprocessing function used by the V3 experiment.

from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# Prepare the same image at the V3 input size.
v3_sample = prepare_image(
    sample_image_path,
    target_size=IMG_SIZE_V3
)

# Apply the official MobileNetV2 preprocessing.
v3_preprocessed_sample = preprocess_input(
    np.expand_dims(v3_sample.copy(), axis=0)
)

print("V3 MOBILENETV2 PREPROCESSING VERIFICATION")
print("=" * 80)

print(f"Input shape:       {v3_sample.shape}")
print(f"Model input shape: {v3_preprocessed_sample.shape}")
print(f"Input pixel range: {v3_sample.min():.4f} to {v3_sample.max():.4f}")
print(
    f"Preprocessed range: "
    f"{v3_preprocessed_sample.min():.4f} to "
    f"{v3_preprocessed_sample.max():.4f}"
)

assert v3_sample.shape == (320, 320, 3)
assert v3_preprocessed_sample.shape == (1, 320, 320, 3)
assert v3_preprocessed_sample.min() >= -1.0
assert v3_preprocessed_sample.max() <= 1.0

print("\nSTATUS: MobileNetV2 preprocessing verified.")
````

### Data Augmentation Strategy

Data augmentation is used to increase the diversity of training examples by applying controlled transformations to training images. The purpose is to improve the model's ability to generalize to variations that may occur in real-world maize leaf images.

Augmentation is applied only to the training dataset. Validation and test images are kept unchanged so that model performance is evaluated on data that has not been artificially modified.

The project uses controlled augmentation across the model experiments:

- **V1 Custom CNN:** Training-only augmentation was used as part of the original CNN training pipeline.
- **V2 Custom CNN:** A moderate augmentation configuration was tested together with a 320 × 320 input size.
- **V3 MobileNetV2:** Moderate training-only augmentation is used with horizontal flipping, small rotations, small zoom changes, and small contrast changes.

1. Random horizontal flip.

4. Random contrast adjustment of approximately 10%.

These transformations are intentionally moderate. Excessive augmentation could alter important disease characteristics such as lesion shape, color, and distribution, potentially making the training examples less representative of actual maize disease patterns.

The same validation and test images remain untouched throughout the experiments, allowing V1, V2, and V3 to be compared using the same evaluation data.

## 4.1 V1 — Custom CNN Baseline

V1 is the project's baseline convolutional neural network developed for four-class maize disease and health classification.

- Blight
- Common_Rust

The V1 pipeline uses pixel normalization by dividing pixel values by 255, training-only data augmentation, and class weights to account for the observed class imbalance.

The trained V1 model and its associated training artifacts are preserved in the project repository and are not overwritten by the subsequent experiments.

````python
# PURPOSE: Inspect the preserved V1 model archive to determine how its architecture and weights are stored.

import zipfile
import json

V1_MODEL_PATH = TRAINING_OUTPUT_DIR / "best_maize_disease_cnn.keras"

assert V1_MODEL_PATH.exists(), f"V1 model not found: {V1_MODEL_PATH}"

print("V1 MODEL ARCHIVE INSPECTION")
print("=" * 80)
print(f"Model path: {V1_MODEL_PATH}")

with zipfile.ZipFile(V1_MODEL_PATH, "r") as z:
    archive_files = z.namelist()

print("\nFiles inside the .keras archive:")
for filename in archive_files:
    print(f"  {filename}")

print("\nSTATUS: V1 model archive inspected.")
````

````python
# PURPOSE: Inspect the original V1 CNN architecture definition so it can be reconstructed exactly if needed.

CNN_MODEL_FILE = PROJECT_ROOT / "cnn_model.py"

assert CNN_MODEL_FILE.exists(), f"V1 architecture file not found: {CNN_MODEL_FILE}"

print("V1 ARCHITECTURE FILE")
print("=" * 80)
print(f"File: {CNN_MODEL_FILE}")

with open(CNN_MODEL_FILE, "r", encoding="utf-8") as f:
    cnn_model_source = f.read()

print("\n--- cnn_model.py ---")
print(cnn_model_source)

print("\nSTATUS: V1 architecture source loaded for inspection.")
````

````python
# PURPOSE: Reconstruct the exact V1 CNN architecture from the project's original cnn_model.py definition.

import tensorflow as tf

IMAGE_SIZE_V1 = 256
NUM_CHANNELS_V1 = 3
NUM_CLASSES_V1 = 4

model_v1_rebuilt = tf.keras.Sequential([

    # Input
    tf.keras.layers.Input(
        shape=(IMAGE_SIZE_V1, IMAGE_SIZE_V1, NUM_CHANNELS_V1)
    ),

    # Convolutional Block 1
    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation="relu",
        padding="same"
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Convolutional Block 2
    tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation="relu",
        padding="same"
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Convolutional Block 3
    tf.keras.layers.Conv2D(
        filters=128,
        kernel_size=(3, 3),
        activation="relu",
        padding="same"
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Convolutional Block 4
    tf.keras.layers.Conv2D(
        filters=256,
        kernel_size=(3, 3),
        activation="relu",
        padding="same"
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Feature aggregation
    tf.keras.layers.GlobalAveragePooling2D(),

    # Classification head
    tf.keras.layers.Dense(
        128,
        activation="relu"
    ),
    tf.keras.layers.Dropout(0.4),

    # Output
    tf.keras.layers.Dense(
        NUM_CLASSES_V1,
        activation="softmax"
    )

], name="maize_disease_cnn")


print("V1 REBUILT ARCHITECTURE")
print("=" * 80)

model_v1_rebuilt.summary()

print("\nSTATUS: V1 architecture reconstructed successfully.")
````

Both stages used the same stratified train, validation, and test split established during data preparation. This ensured that the V3 results could be compared fairly with V1 and V2.

The purpose of V3 was not simply to obtain a higher accuracy, but to determine whether transfer learning and controlled fine-tuning could improve generalization and reduce the classification difficulty observed between Blight and Gray_Leaf_Spot.

## Confidence and Uncertainty Evaluation

Overall classification accuracy does not indicate how confident the model is in individual predictions. For this project, prediction confidence is therefore considered alongside classification performance.

The confidence analysis evaluates whether predictions can be separated into higher-confidence accepted predictions and lower-confidence uncertain predictions.

The project's confidence policy uses a **70% acceptance threshold**. Predictions with confidence below this threshold are treated as uncertain rather than being presented as confident disease identifications.

This approach is particularly important for the Blight and Gray_Leaf_Spot classes because these diseases showed the greatest remaining visual overlap during error analysis.

The confidence mechanism is intended as a safety layer rather than a guarantee of correctness. A high-confidence prediction can still be wrong, while an uncertain prediction may still correspond to the correct class.

Therefore, the system communicates predictions as indications rather than confirmed diagnoses and uses uncertainty handling to reduce the risk of presenting low-confidence predictions as reliable diagnoses.

## Confidence Threshold Findings

The confidence analysis demonstrates a trade-off between prediction coverage and accepted-prediction accuracy.

At the selected **70% confidence threshold**, the system accepts 556 of the 628 test predictions, corresponding to **88.54% coverage**. Of these accepted predictions, 541 are correct and 15 are incorrect, giving an accepted accuracy of **97.30%**.

Increasing the confidence threshold further improves the accuracy of accepted predictions but reduces coverage. For example, a 95% threshold produces **99.78% accepted accuracy**, but only **71.82% of the test predictions are accepted**.

The 70% threshold was therefore retained as a practical balance between accepting useful predictions and identifying predictions that should be treated as uncertain.

Importantly, confidence should not be interpreted as a probability of guaranteed correctness. The uncertainty mechanism is a risk-reduction measure that helps prevent low-confidence predictions from being presented as reliable diagnoses.

The remaining accepted errors also demonstrate why the system requires appropriate caution in communicating predictions, particularly for visually similar disease classes such as Blight and Gray_Leaf_Spot.
